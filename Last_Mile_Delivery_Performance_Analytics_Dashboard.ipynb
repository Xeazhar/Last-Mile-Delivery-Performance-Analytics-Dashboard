```python
import pandas as pd
import numpy as np
from sklearn.cluster import DBSCAN
from scipy.stats import zscore
from fastdtw import fastdtw
from tabulate import tabulate

# --- Load and preprocess data ---
df = pd.read_csv("/content/data6.csv", on_bad_lines='skip')

# Convert columns that should be numeric to float, coercing errors to NaN
df['current_number_of_passengers'] = pd.to_numeric(df['current_number_of_passengers'], errors='coerce')
df['passenger_pickup'] = pd.to_numeric(df['passenger_pickup'], errors='coerce')
df['passenger_dropoff'] = pd.to_numeric(df['passenger_dropoff'], errors='coerce')

# Parse coordinates to lat, lon floats
df['coordinates'] = (
    df['coordinates']
    .astype(str)
    .str.replace(r'[\[\]\(\)]', '', regex=True)
    .str.split(',')
)

df['lon'] = df['coordinates'].apply(lambda x: float(x[0].strip()) if len(x) > 1 and x[0].strip() else np.nan)
df['lat'] = df['coordinates'].apply(lambda x: float(x[1].strip()) if len(x) > 1 and x[1].strip() else np.nan)
df.dropna(subset=['lat', 'lon'], inplace=True)

# Convert timestamp to pandas datetime if not already
df['timestamp'] = pd.to_datetime(df['timestamp'])

# Generate list of all unique trip IDs in the dataframe
trip_ids = df['trip_code'].dropna().unique().tolist()

print(f"Number of unique trips detected: {len(trip_ids)}")

# Filter dataframe to only those trips (should be all, but for safety)
df = df[df['trip_code'].isin(trip_ids)].copy()

# ----------------------
# 1. Spatial Anomaly Detection with DBSCAN on pickups and dropoffs
# ----------------------

def detect_spatial_anomalies(df, status_keyword, eps=0.002, min_samples=3):
    subset = df[df['status'].str.contains(status_keyword, case=False, na=False)].copy()
    if subset.empty:
        return pd.DataFrame(), pd.DataFrame(), None
    coords = subset[['lat', 'lon']].to_numpy()
    clustering = DBSCAN(eps=eps, min_samples=min_samples).fit(coords)
    subset['cluster'] = clustering.labels_
    anomalies = subset[subset['cluster'] == -1]
    return anomalies, subset, clustering

pickup_anomalies, pickup_clusters, pickup_clustering = detect_spatial_anomalies(df, 'board')
dropoff_anomalies, dropoff_clusters, dropoff_clustering = detect_spatial_anomalies(df, 'drop')

print(f"\n--- Spatial Anomaly Detection ---")
if pickup_clustering:
    print(f"DBSCAN parameters for Pickups: eps={pickup_clustering.eps}, min_samples={pickup_clustering.min_samples}")
else:
    print("No pickup data to perform DBSCAN.")
if dropoff_clustering:
    print(f"DBSCAN parameters for Dropoffs: eps={dropoff_clustering.eps}, min_samples={dropoff_clustering.min_samples}")
else:
    print("No dropoff data to perform DBSCAN.")

print(f"Pickup anomalies (noise points, outside clusters): {len(pickup_anomalies)}")
print(f"Dropoff anomalies (noise points, outside clusters): {len(dropoff_anomalies)}")

# ----------------------
# 2. Temporal Anomaly Detection: z-score on pickups per hour
# ----------------------

# Extract hour of day
df['hour'] = df['timestamp'].dt.hour

# Count pickups per hour
hourly_pickups = df[df['status'].str.contains('board', case=False)].groupby('hour').size()

print("\n--- Temporal Anomaly Detection ---")
if not hourly_pickups.empty:
    # Compute z-scores
    hourly_zscores = zscore(hourly_pickups)

    mean_pickups = np.mean(hourly_pickups)
    std_pickups = np.std(hourly_pickups)

    print(f"Mean pickups per hour: {mean_pickups:.2f}")
    print(f"Standard deviation of pickups per hour: {std_pickups:.2f}")
    print(f"Using z-score threshold of |z| > 2 to detect unusual pickup counts per hour")

    # Flag hours with |z| > 2 as anomalies
    anomalous_hours = hourly_pickups[(np.abs(hourly_zscores) > 2)]
    anomalous_hours_z = hourly_zscores[(np.abs(hourly_zscores) > 2)]

    if len(anomalous_hours) > 0:
        temporal_anomaly_table = pd.DataFrame({
            'Hour of Day': anomalous_hours.index,
            'Pickup Count': anomalous_hours.values,
            'Z-Score': anomalous_hours_z.tolist() # Convert numpy array to list
        })
        print(tabulate(temporal_anomaly_table, headers='keys', tablefmt='psql'))
    else:
        print("No temporal anomalies detected.")

    # Optional: Show all hours with counts and z-scores for context
    all_hours_table = pd.DataFrame({
        'Hour of Day': hourly_pickups.index,
        'Pickup Count': hourly_pickups.values,
        'Z-Score': hourly_zscores
    })
    print("\nAll hourly pickups with z-scores:")
    print(tabulate(all_hours_table, headers='keys', tablefmt='psql'))

else:
    print("No hourly pickup data to perform temporal anomaly detection.")


# ----------------------
# 3. Structural Anomaly Detection: route similarity with DTW
# ----------------------

# Group data by trip_code and sort by timestamp
trips = []
for trip_id in trip_ids:
    trip_data = df[df['trip_code'] == trip_id].sort_values('timestamp')
    coords = trip_data[['lat', 'lon']].to_numpy()
    if len(coords) > 1:
        trips.append((trip_id, coords))

print("\n--- Structural Anomaly Detection ---")
if trips:
    # Compute DTW distance between all pairs of routes using fastdtw
    def dtw_distance(route1, route2):
        distance, _ = fastdtw(route1, route2, dist=lambda x, y: np.linalg.norm(x - y)) # Euclidean distance
        return distance

    # Compute average distance of each trip to all others
    avg_distances = []
    for i, (trip_id_i, route_i) in enumerate(trips):
        dists = []
        for j, (trip_id_j, route_j) in enumerate(trips):
            if i != j:
                dists.append(dtw_distance(route_i, route_j))
        avg_dist = np.mean(dists) if dists else 0
        avg_distances.append((trip_id_i, avg_dist))

    # Calculate threshold: mean + 1.5 * std
    dist_values = [dist for _, dist in avg_distances]
    threshold = np.mean(dist_values) + 1.5 * np.std(dist_values) if dist_values else 0

    # Find typical route: trip with smallest average distance
    typical_trip_id, typical_avg_dist = min(avg_distances, key=lambda x: x[1]) if avg_distances else (None, None)

    if typical_trip_id:
        print(f"Typical route is trip_id {typical_trip_id} with average DTW distance {typical_avg_dist:.4f}")
        print(f"Using DTW average distance threshold: mean + 1.5 * std = {threshold:.4f}")
        # Identify structural anomalies
        structural_anomalies = [(tid, dist) for tid, dist in avg_distances if dist > threshold]

        if structural_anomalies:
            structural_table = pd.DataFrame(structural_anomalies, columns=['Trip ID', 'Avg DTW Distance'])
            structural_table['Distance Above Threshold'] = structural_table['Avg DTW Distance'] - threshold
            print("Trips flagged as structural anomalies due to significantly different routes:")
            print(tabulate(structural_table, headers='keys', tablefmt='psql'))
        else:
            print("No structural anomalies detected.")
    else:
        print("No trips with enough data to perform structural anomaly detection.")
else:
    print("No trips found in the filtered data to perform structural anomaly detection.")

# ----------------------
# Display sample pickup and dropoff anomalies with explanations
# ----------------------

cols_order = ['trip_code', 'timestamp', 'status', 'lat', 'lon']

print("\nSample Pickup Anomalies (noise points outside spatial clusters):")
if not pickup_anomalies.empty:
    print(tabulate(pickup_anomalies[cols_order].head(5), headers='keys', tablefmt='psql'))
else:
    print("No pickup anomalies detected to display.")

print("\nSample Dropoff Anomalies (noise points outside spatial clusters):")
if not dropoff_anomalies.empty:
    print(tabulate(dropoff_anomalies[cols_order].head(5), headers='keys', tablefmt='psql'))
else:
    print("No dropoff anomalies detected to display.")
```

```python
import pandas as pd
import folium
from sklearn.cluster import DBSCAN
from scipy.spatial.distance import cdist
import numpy as np
from folium import MacroElement
from jinja2 import Template

# --- Load dataset ---
df = pd.read_csv("/content/data4.csv", on_bad_lines='skip')

# --- Clean and parse coordinates ---
df['coordinates'] = (
    df['coordinates']
    .astype(str)
    .str.replace(r'[\[\]\(\)]', '', regex=True)
    .str.split(',')
)

df['lon'] = df['coordinates'].apply(lambda x: float(x[0].strip()) if len(x) > 1 else None)
df['lat'] = df['coordinates'].apply(lambda x: float(x[1].strip()) if len(x) > 1 else None)
df = df.dropna(subset=['lat', 'lon'])

# --- Filter for specified trip_ids ---
trip_ids = [
    "0c2999cd-1caa-4d70-86b8-09fef98bc589",
    "154b49a7-9a89-47f6-973a-a192bc3e02c5",
    "11920ad7-cb7e-4c5e-b6e1-3b35b63f63f2",
    "01d5f40c-40fe-441a-9f0f-f2b3c8cd9a87",
    "154b49a7-9a89-47f6-973a-a192bc3e02c5",
    "11920ad7-cb7e-4c5e-b6e1-3b35b63f63f2",
    "01d5f40c-40fe-441a-9f0f-f2b3c8cd9a87",
    "dde7bce3-3ead-424f-a90d-d55e323f82ba",
    "ab43e432-cd85-4c59-be4e-3e313a1d0ff9",
    "49fc9847-6969-4754-a950-b46c44976f1e",
    "a43d9b7b-4eb1-4320-9108-af28684c25f9",
    "154b49a7-9a89-47f6-973a-a192bc3e02c5"
]

df = df[df['trip_code'].isin(trip_ids)].copy()

# --- Filter pickups and dropoffs ---
pickup_df = df[df['status'].str.contains('board', case=False, na=False)].copy()
dropoff_df = df[df['status'].str.contains('drop', case=False, na=False)].copy()

# --- DBSCAN clustering function ---
def cluster_points(df, eps=0.002, min_samples=3):
    coords = df[['lat', 'lon']]
    db = DBSCAN(eps=eps, min_samples=min_samples).fit(coords)
    df['cluster'] = db.labels_
    return df

# --- Cluster pickups and dropoffs separately ---
pickup_df = cluster_points(pickup_df)
dropoff_df = cluster_points(dropoff_df)

# --- Initialize map centered on average location ---
center_lat = df['lat'].mean()
center_lon = df['lon'].mean()
m = folium.Map(location=[center_lat, center_lon], zoom_start=14)

# --- Helper to scale circle radius based on cluster size ---
def scale_radius(count, min_r=6, max_r=28, max_count=None):
    if max_count is None or max_count == 0:
        return min_r
    return min_r + (max_r - min_r) * (count / max_count)

# --- Get medoid point of a cluster ---
def get_medoid_point(group):
    coords = group[['lat', 'lon']].to_numpy()
    dist_matrix = cdist(coords, coords, metric='euclidean')
    dist_sums = dist_matrix.sum(axis=1)
    medoid_idx = np.argmin(dist_sums)
    medoid_point = coords[medoid_idx]
    return medoid_point[0], medoid_point[1]

# --- Plot clusters and noise with medoid as cluster location ---
def plot_clusters(df, cluster_color, noise_color, label):
    clusters = df[df['cluster'] != -1].groupby('cluster')
    max_count = clusters.size().max() if not clusters.size().empty else 1

    # Plot cluster centers (medoid) sized by cluster count
    for cluster_id, group in clusters:
        lat_medoid, lon_medoid = get_medoid_point(group)
        radius = scale_radius(len(group), max_count=max_count)

        folium.CircleMarker(
            location=[lat_medoid, lon_medoid],
            radius=radius,
            color=cluster_color,
            fill=True,
            fill_opacity=0.6,
            popup=f"{label} Cluster {cluster_id} ({len(group)} points)"
        ).add_to(m)

    # Plot noise points as anomalies
    noise = df[df['cluster'] == -1]
    for _, row in noise.iterrows():
        folium.CircleMarker(
            location=[row['lat'], row['lon']],
            radius=6,
            color=noise_color,
            fill=True,
            fill_opacity=0.9,
            popup=f"{label} Anomaly (noise point)"
        ).add_to(m)

# --- Plot pickup and dropoff clusters ---
plot_clusters(pickup_df, cluster_color='blue', noise_color='red', label='Pickup')
plot_clusters(dropoff_df, cluster_color='green', noise_color='orange', label='Dropoff')

# --- Add routes with very low opacity ---
for trip_id in trip_ids:
    trip_data = df[df['trip_code'] == trip_id].sort_values('timestamp')
    coords = list(zip(trip_data['lat'], trip_data['lon']))
    if len(coords) > 1:
        folium.PolyLine(coords, color='gray', weight=2, opacity=0.2, tooltip=f"Route: {trip_id[:6]}").add_to(m)

# --- Add legend ---
legend_html = '''
<div style="
    position: fixed;
    top: 10px; right: 10px;
    z-index:9999;
    background-color: white;
    padding: 10px;
    border: 2px solid grey;
    border-radius: 6px;
    box-shadow: 3px 3px 6px rgba(0,0,0,0.3);
    font-size: 14px;
    ">
    <b>Legend</b><br>
    <i style="background: blue; width: 12px; height: 12px; border-radius: 50%; display: inline-block;"></i> Pickup Cluster<br>
    <i style="background: green; width: 12px; height: 12px; border-radius: 50%; display: inline-block;"></i> Dropoff Cluster<br>
    <i style="background: red; width: 12px; height: 12px; border-radius: 50%; display: inline-block;"></i> Pickup Anomaly<br>
    <i style="background: orange; width: 12px; height: 12px; border-radius: 50%; display: inline-block;"></i> Dropoff Anomaly<br>
    <i style="background: gray; width: 20px; height: 3px; display: inline-block; margin-top: 6px;"></i> Route Line
</div>
'''

m.get_root().html.add_child(folium.Element(legend_html))

# Show the map
m
```

```python
import pandas as pd
import folium
from IPython.display import display

# --- Load dataset ---
df = pd.read_csv("/content/data6.csv", on_bad_lines='skip')

# --- Clean and parse coordinates ---
df['coordinates'] = (
    df['coordinates']
    .astype(str)
    .str.replace(r'[\[\]\(\)]', '', regex=True)
    .str.split(',')
)

df['lon'] = df['coordinates'].apply(lambda x: float(x[0].strip()) if len(x) > 1 else None)
df['lat'] = df['coordinates'].apply(lambda x: float(x[1].strip()) if len(x) > 1 else None)

df = df.dropna(subset=['lat', 'lon'])

trip_ids = [
    "11920ad7-cb7e-4c5e-b6e1-3b35b63f63f2",
    "0c2999cd-1caa-4d70-86b8-09fef98bc589"
]

colors = ['green', 'purple', 'orange', 'blue']

first_point = None
for trip_id in trip_ids:
    trip_data = df[df['trip_code'] == trip_id]
    if not trip_data.empty:
        first_point = trip_data.iloc[0]
        break

if first_point is None:
    print("No data found for the given trip codes.")
else:
    m = folium.Map(location=[first_point['lat'], first_point['lon']], zoom_start=14)

    for idx, trip_id in enumerate(trip_ids):
        trip_data = df[df['trip_code'] == trip_id].copy()

        if trip_data.empty:
            print(f"No data found for trip_code {trip_id}")
            continue

        coords = list(zip(trip_data['lat'], trip_data['lon']))
        passengers = trip_data['current_number_of_passengers'].astype(float).tolist()

        # Draw route line WITHOUT arrows
        folium.PolyLine(coords, color=colors[idx % len(colors)], weight=5, opacity=0.7, tooltip=f"Route: {trip_id[:6]}").add_to(m)

        # Add pickup/drop-off markers
        for i in range(1, len(passengers)):
            prev_pax = passengers[i - 1]
            curr_pax = passengers[i]
            point = coords[i]

            if curr_pax > prev_pax:
                folium.CircleMarker(
                    location=point,
                    radius=5,
                    color="blue",
                    fill=True,
                    fill_opacity=0.8,
                    popup=f"Pickup | Trip: {trip_id[:6]} | Pax: {curr_pax}"
                ).add_to(m)
            elif curr_pax < prev_pax:
                folium.CircleMarker(
                    location=point,
                    radius=5,
                    color="red",
                    fill=True,
                    fill_opacity=0.8,
                    popup=f"Drop-off | Trip: {trip_id[:6]} | Pax: {curr_pax}"
                ).add_to(m)

        # Start point marker (green)
        folium.Marker(
            location=coords[0],
            icon=folium.Icon(color='green', icon='play'),
            popup=f"Start | Trip: {trip_id[:6]}"
        ).add_to(m)

        # End point marker (red)
        folium.Marker(
            location=coords[-1],
            icon=folium.Icon(color='red', icon='stop'),
            popup=f"End | Trip: {trip_id[:6]}"
        ).add_to(m)

    display(m)
```
